# Markov Decision Process



Markov Decision Process (MDP) é definido por uma tupla (S, A, T, R), onde:

- S -> State: lista de todos os estados possíveis do problema
- A -> Actions: Lista de todas as ações que o agente pode tomar
- T -> Transitions: Função das probabilidades de transição entre os estados
- R -> Reinforcement: Função de reforço imediato


## Modelagem do Pacman

Considerando o jogo classico PacMan, podemos pensar em duas modelagens distintas: uma determinística e outra probabilística.
Vamos começar pela determinística, que é mais simples.

### Modelagem Determinística

```
S = { 
    - posição do pacman;
    - direção do pacman;
    - estado do pacman (superpoder ou normal);
    - posição dos fantasmas;
    - direção dos fantasmas;
    - posição das pastilhas;
    - posição das superpastilhas
}
```

```
A = {
    - subir
    - descer
    - esquerda
    - direita
    - manter direção atual
}
```

```
R = {
    -  +1 para comer uma pastilha normal
    -  +5 para comer uma super pastilha 
    - +10 para comer um fantasma
    - +50 vitória
    - -50 derrota
    -   0 para todo o resto
}
```

T -> Para modelagem determinística, a partir do estado atual deve ser possível determinar o próximo estado. Isso implica que a próxima posição dos fantasmas deve ser conhecida, ou seja, a regra de movimentação dos fantasmas é fixa, conhecida, e não tem nenhum tipo de aleatoriedade. Isso pode depender de como o jogo está programado e poderíamos adotar essa modelagem determinística assumindo essa simplificação proposta.

### Modelagem Probabilística

Para uma abordagem probabilística, é necessário considerar a probabilidade dos movimentos de cada fantasma no jogo. Como eles são controlados pelo jogo, e não pelo agente, devemos conhecer qual a probabilidade de cada fantasma ir para uma determinada posição no labirinto. Outra forma poderia ser conhecer a probabilidade de movimentação de cada fantasma, associada com a probabilidade dele conseguir realizar tal movimento: as vezes os fantasmas mudam de direção de forma aleatória. Dessa forma poderiamos modelar a função T e predizer o próximo estado do jogo. Para as demais funções, acredito que podemos adotar a mesma modelagem determinística. 

### Outras Considerações

Primeiro sobre a função de reforço: os números são sugestões preliminares. Acredito que devamos treinar o agente com tais valores e observar os resultados obtidos. Em função disso podemos fazer ajustes nas recompensas e voltar a treinar o agente, realizando uma espécie de "otimização" dos hiperparametros do modelo.

Outro ponto, nós não modelamos a quantidade de vidas do pacman no problema. Nem a possibilidade de ganhar vidas durante o jogo. Isso é uma simplificação razoável, mas poderíamos querer modelar também essa característica do jogo. Isso permitiría ao processo de treinamento conseguir estratégias que considerassem o sacrifício em determinada situação em pró de conseguir fechar determinado estágio num segundo momento. Mas seria certamente uma modelagem muito mais complexa.

## Modelagem do Xadrez

O jogo de Xadrez por definição possui característica determinística: dado o estado atual e o jogador de turno, é possível determinar completamente o próximo estado a partir disso e da ação escolhida. Então nossa função T seria algo simples, como:

```
T = { 
  100% -> conforme as regras do jogo,
    0% -> para qualquer movimento inválido
}
```

A função S também é relativamente simples. Podemos defini-la como:

```
S = {
  - Posição de todas as peças no tabuleiro
  - Roque executado por cada jogador (ou não executado), pois é um movimento especial que não pode ser inteiramente determinado apenas pela posição atual do tabuleiro
  - Qual o jogador de turno
}
```

A função A também é relativamente simples, embora extensa. Eu modelaria ela como sendo:

```
A = {
  Tipo de peça: conjunto de casas possíveis de serem atingidas pelo tipo de peça durante o jogo
}
```

Ja a função de reforço, diferente do pacman, me parece bastante complexa. Quase uma heurística que deva ser determinada por alguns elementos, como por exemplo:

- Valor relativo de cada peça
- Conjunto das dominâncias sobre cada casa do tabuleiro (quantidade de peças que atacam determinada cada subtraído da quantidade de peças que defendem determinada cada)
- Valor relativo de cada casa no tabuleiro (o centro do tabuleiro costuma valer mais que os cantos pois permitem atacar mais casas)
- Valor das peças adversárias posicionadas sobre cada casa dominada por voce
- Valor das peças suas posicionadas sobre cada casa dominada pelo adversário

No estudo de Xadrez, o valor das peças é tipicamente fixo. Peões valem 1, bispos e cavalos valem 3 (embora alguns autores defendas que existe diferenças em determinadas situações), torres valem 5, rainha vale 10, e normalmente não estabelecem valor nominal para o Rei, por ser a peça chave mais importante no tabuleiro. Para nossa modelagem, entretanto, vamos supor que o valor seja muito superior a todas as demais peças, por exemplo 500.

Nossa função de reforço deveria ser testada e ajustada. Essa é apenas minha intuição a partir do que conheço a respeito do jogo.
